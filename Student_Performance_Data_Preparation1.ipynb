```python

# Correct million-row dataset
df_million = pd.read_csv("StudentsPerformance.csv")

# Correct exams dataset
df_exams = pd.read_csv("StudentPerformance.csv")

```

```python


print("EXAMS DATASET")
print("Missing values:", df_exams.isnull().sum().sum())
print("Duplicate rows:", df_exams.duplicated().sum())

print("\nMILLION DATASET")
print("Missing values:", df_million.isnull().sum().sum())
print("Duplicate rows:", df_million.duplicated().sum())
print("Unique student IDs:", df_million["student_id"].nunique())
```

```python

# ==================================
# EXAMS DATASET: 70 / 15 / 15
# ==================================

exams_train, exams_temp = train_test_split(
    df_exams,
    test_size=0.30,
    random_state=42,
    shuffle=True
)

exams_validation, exams_test = train_test_split(
    exams_temp,
    test_size=0.50,
    random_state=42,
    shuffle=True
)


# ==================================
# MILLION DATASET: 70 / 15 / 15
# ==================================

million_train, million_temp = train_test_split(
    df_million,
    test_size=0.30,
    random_state=42,
    shuffle=True,
    stratify=df_million["grade"]
)

million_validation, million_test = train_test_split(
    million_temp,
    test_size=0.50,
    random_state=42,
    shuffle=True,
    stratify=million_temp["grade"]
)

print("ALL DATASETS SPLIT SUCCESSFULLY!")
```

```python

print("EXAMS DATASET")
print("Train:", len(exams_train))
print("Validation:", len(exams_validation))
print("Test:", len(exams_test))

print("\nMILLION DATASET")
print("Train:", len(million_train))
print("Validation:", len(million_validation))
print("Test:", len(million_test))
```

```python

import os

os.makedirs("parquet_output", exist_ok=True)

print("Parquet output folder created successfully!")
```

```python

# EXAMS DATASET

exams_train.to_parquet(
    "parquet_output/exams_train.parquet", index=False
)

exams_validation.to_parquet(
    "parquet_output/exams_validation.parquet", index=False
)

exams_test.to_parquet(
    "parquet_output/exams_test.parquet", index=False
)


# MILLION DATASET

million_train.to_parquet(
    "parquet_output/million_train.parquet", index=False
)

million_validation.to_parquet(
    "parquet_output/million_validation.parquet", index=False
)

million_test.to_parquet(
    "parquet_output/million_test.parquet", index=False
)

print("SUCCESS! ALL SIX PARQUET FILES CREATED!")
```
